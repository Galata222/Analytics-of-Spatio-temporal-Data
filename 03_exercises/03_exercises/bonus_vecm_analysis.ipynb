# Bonus: VAR/VECM analýza energetických, kovových a technologických dát

Notebook analyzuje datasety `energy.csv`, `Precious_Metals.csv` a `Technology.csv`.

## Postup

1. ADF test v úrovniach a v prvých diferenciách.
2. Johansenov test a určenie kointegračného ranku.
3. Výber datasetu vhodného pre VECM.
4. Výber počtu lagov podľa AIC.
5. Odhad VECM a rolling-origin predikcia na posledne tri mesiace.
6. Výpočet MAE a RMSE a graf skutočných verzus predikovaných hodnôt.

VECM používa pôvodné urovne premenných, pretože error-correction člen modeluje dlhodobý kointegracny vzťah. Krátkodobá dynamika je v modeli reprezentovaná prvymi diferenciami.


```python


from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.vector_ar.vecm import VECM, coint_johansen, select_order


BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR / "data"
DATASETS = {
    "energy": DATA_DIR / "energy.csv",
    "Precious_Metals": DATA_DIR / "Precious_Metals.csv",
    "Technology": DATA_DIR / "Technology.csv",
}


def adf_pvalue(series: pd.Series) -> float:
    """Vrati p-hodnotu ADF testu s vyberom lagov podla AIC."""
    return float(adfuller(series.dropna(), autolag="AIC")[1])


def johansen_rank(levels: pd.DataFrame) -> tuple[int, object]:
    """Urci rank podla prvej nezamietnutej hypotezy stopovej statistiky na 5 % hladine."""
    result = coint_johansen(levels, det_order=0, k_ar_diff=1)
    for rank, (trace_stat, critical_value) in enumerate(
        zip(result.lr1, result.cvt[:, 1])
    ):
        if trace_stat <= critical_value:
            return rank, result
    return levels.shape[1], result


def mean_absolute_error(actual: pd.Series, predicted: pd.Series) -> float:
    return float(np.mean(np.abs(actual.to_numpy() - predicted.to_numpy())))


def root_mean_squared_error(actual: pd.Series, predicted: pd.Series) -> float:
    errors = actual.to_numpy() - predicted.to_numpy()
    return float(np.sqrt(np.mean(errors**2)))


def load_dataset(path: Path) -> pd.DataFrame:
    if not path.exists():
        raise FileNotFoundError(f"Dataset sa nenašiel: {path}")
    frame = pd.read_csv(path, parse_dates=["Date"]).set_index("Date").sort_index()
    levels = frame.select_dtypes(include="number").dropna()
    if levels.shape[1] < 2:
        raise ValueError(f"{path.name} musi obsahovat aspon dva ciselne casove rady.")
    return levels


def main() -> None:
    results = {}
    summary_rows = []

    for name, path in DATASETS.items():
        levels = load_dataset(path)
        rank, johansen = johansen_rank(levels)
        level_pvalues = {column: adf_pvalue(levels[column]) for column in levels}
        difference_pvalues = {
            column: adf_pvalue(levels[column].diff()) for column in levels
        }
        results[name] = {
            "levels": levels,
            "rank": rank,
            "johansen": johansen,
            "level_pvalues": level_pvalues,
            "difference_pvalues": difference_pvalues,
        }
        summary_rows.append(
            {
                "dataset": name,
                "observations": len(levels),
                "variables": levels.shape[1],
                "kointegracny rank r": rank,
                "ADF p-hodnoty v úrovniach": ", ".join(
                    f"{column}: {value:.4f}"
                    for column, value in level_pvalues.items()
                ),
                "ADF p-hodnoty v prvých diferenciách": ", ".join(
                    f"{column}: {value:.4f}"
                    for column, value in difference_pvalues.items()
                ),
                "odporucany model": "VECM" if rank > 0 else "VAR v prvych diferenciach",
            }
        )

        trace_table = pd.DataFrame(
            {
                "H0": [f"r <= {i}" for i in range(len(johansen.lr1))],
                "stopova statistika": johansen.lr1,
                "kriticka hodnota 5 %": johansen.cvt[:, 1],
                "zamietnutie na 5 %": johansen.lr1 > johansen.cvt[:, 1],
            }
        )
        print(f"\n{name}: Johansenov test stopovej statistiky")
        print(trace_table.round(4).to_string(index=False))

    summary = pd.DataFrame(summary_rows).set_index("dataset")
    print("\nSuhrn klasifikacie")
    print(summary.to_string())
    print(
        "\nInterpretacia: r je pocet nezavislych stacionarnych "
        "kombinacii nestacionarnych casovych radov v urovniach."
    )
    print(
        "Kladny rank znamena, ze VECM moze zachytit kratkodobu "
        "dynamiku aj dlhodobý korekciu rovnovazneho vzťahu."
    )

    vecm_candidates = [name for name, item in results.items() if item["rank"] > 0]
    if len(vecm_candidates) != 1:
        raise ValueError(
            f"Ocakaval sa prave jeden kandidat pre VECM, najdene: {vecm_candidates}"
        )

    vecm_name = vecm_candidates[0]
    levels = results[vecm_name]["levels"]
    rank = results[vecm_name]["rank"]

    # Posledne tri kalendarne mesiace su testovacie obdobie; kazda predikcia je o krok dopredu.
    test_start = levels.index.max() - pd.DateOffset(months=3)
    train = levels.loc[levels.index < test_start]
    test = levels.loc[levels.index >= test_start]
    if len(train) < 30 or test.empty:
        raise ValueError("Vybrany dataset je prilis kratky na pozadovane rozdelenie.")

    lag_order = select_order(train, maxlags=10, deterministic="ci")
    k_ar_diff = lag_order.aic
    if k_ar_diff is None or not np.isfinite(k_ar_diff):
        raise ValueError("AIC neurcilo platny pocet lagov pre VECM.")
    k_ar_diff = int(k_ar_diff)

    fitted_model = VECM(
        train,
        k_ar_diff=k_ar_diff,
        coint_rank=rank,
        deterministic="ci",
    ).fit()

    predictions = []
    for date in test.index:
        history = levels.loc[levels.index < date]
        rolling_fit = VECM(
            history,
            k_ar_diff=k_ar_diff,
            coint_rank=rank,
            deterministic="ci",
        ).fit()
        predictions.append(rolling_fit.predict(steps=1)[0])

    predicted = pd.DataFrame(predictions, index=test.index, columns=levels.columns)
    metrics = pd.DataFrame(
        {
            "MAE": {
                column: mean_absolute_error(test[column], predicted[column])
                for column in levels.columns
            },
            "RMSE": {
                column: root_mean_squared_error(test[column], predicted[column])
                for column in levels.columns
            },
        }
    )

    print(f"\nDataset pre VECM: {vecm_name}")
    print(f"Kointegracny rank: r = {rank}")
    print(f"Treningove obdobie: {train.index.min().date()} to {train.index.max().date()}")
    print(f"Testovacie obdobie: {test.index.min().date()} to {test.index.max().date()}")
    print(f"Pocet diferencovanych lagov k_ar_diff podla AIC: {k_ar_diff}")
    print("\nMetriky rolling-origin jednokrokovej predikcie")
    print(metrics.round(4).to_string())

    fig, axes = plt.subplots(
        len(levels.columns),
        1,
        figsize=(12, 3 * len(levels.columns)),
        sharex=True,
    )
    axes = np.atleast_1d(axes)
    for axis, column in zip(axes, levels.columns):
        axis.plot(test.index, test[column], label="Skutočná hodnota")
        axis.plot(test.index, predicted[column], label="Predikcia")
        axis.set_ylabel(column)
        axis.grid(alpha=0.3)
        axis.legend()
    axes[-1].set_xlabel("Dátum")
    fig.suptitle(f"Rolling-origin VECM predikcia: {vecm_name}")
    fig.tight_layout()
    output_path = BASE_DIR / "energy_vecm_rolling_forecast.png"
    fig.savefig(output_path, dpi=160, bbox_inches="tight")
    print(f"\nGraf ulozeny do: {output_path}")
    plt.show()


main()
```
